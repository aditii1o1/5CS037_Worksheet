# Data Understanding, Analysis, and Preparation

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Load the dataset
data = pd.read_csv('student.csv')

# Observe the dataset
print("Top 5 rows of the dataset:")
print(data.head())

print("\nBottom 5 rows of the dataset:")
print(data.tail())

# Dataset information
print("\nDataset Information:")
print(data.info())

# Descriptive statistics
print("\nDescriptive Statistics:")
print(data.describe())

# Split data into features (X) and target (Y)
X = data[['Math', 'Reading']].values  # Features: Math and Reading marks
Y = data['Writing'].values            # Target: Writing marks

# Split data into training and testing sets (80-20 split)
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

print("\nData split completed.")
```

#  Build a Cost Function

```python
import numpy as np

def cost_function(X, Y, W):
    """
    Compute the Mean Squared Error.
    """
    m = len(Y)  # Number of samples
    predictions = np.dot(X, W)
    errors = predictions - Y
    cost = (1 / (2 * m)) * np.sum(errors ** 2)
    return cost

# Test the cost function with sample data
X_test = np.array([[1, 2], [3, 4], [5, 6]])
Y_test = np.array([3, 7, 11])
W_test = np.array([1, 1])

cost = cost_function(X_test, Y_test, W_test)
print("\nCost for the test data:", cost)
```

# Gradient Descent for Simple Linear Regression



```python
def gradient_descent(X, Y, W, alpha, iterations):
    """
    Perform gradient descent to optimize weights.
    """
    m = len(Y)  # Number of samples
    cost_history = []  # To store cost at each iteration

    for i in range(iterations):
        predictions = np.dot(X, W)
        errors = predictions - Y
        gradients = (1 / m) * np.dot(X.T, errors)
        W -= alpha * gradients  # Update weights
        cost = cost_function(X, Y, W)
        cost_history.append(cost)
    
    return W, cost_history

# Test gradient descent with sample data
np.random.seed(0)
X = np.random.rand(100, 3)  # 100 samples, 3 features
Y = np.random.rand(100)
W = np.random.rand(3)       # Initial weights
alpha = 0.01
iterations = 1000

W_optimized, cost_history = gradient_descent(X, Y, W, alpha, iterations)
print("\nOptimized Weights:", W_optimized)
print("Cost History (First 10):", cost_history[:10])
```

# Evaluate the Model
## RMSE

```python
def rmse(Y, Y_pred):
    """
    Compute Root Mean Squared Error.
    """
    return np.sqrt(np.mean((Y - Y_pred) ** 2))

# Test RMSE
Y_pred = np.dot(X_test, W_test)
rmse_value = rmse(Y_test, Y_pred)
print("\nRMSE for test data:", rmse_value)
```

## R squared

```python
def r2(Y, Y_pred):
    """
    Compute R-squared value.
    """
    mean_y = np.mean(Y)
    ss_tot = np.sum((Y - mean_y) ** 2)
    ss_res = np.sum((Y - Y_pred) ** 2)
    return 1 - (ss_res / ss_tot)

# Test R-squared
r2_value = r2(Y_test, Y_pred)
print("\nR-squared value for test data:", r2_value)
```

# Main Function 

```python
def main():
    # Load dataset
    data = pd.read_csv('student.csv')
    X = data[['Math', 'Reading']].values
    Y = data['Writing'].values

    # Split dataset into training and testing sets
    X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

    # Initialize weights, learning rate, and iterations
    W = np.zeros(X_train.shape[1])
    alpha = 0.0001
    iterations = 1000

    # Perform gradient descent
    W_optimized, cost_history = gradient_descent(X_train, Y_train, W, alpha, iterations)

    # Predictions
    Y_pred = np.dot(X_test, W_optimized)

    # Evaluation
    model_rmse = rmse(Y_test, Y_pred)
    model_r2 = r2(Y_test, Y_pred)

    # Output results
    print("\nFinal Weights:", W_optimized)
    print("RMSE on Test Set:", model_rmse)
    print("R-Squared on Test Set:", model_r2)

# Execute main function
if __name__ == "__main__":
    main()
```
